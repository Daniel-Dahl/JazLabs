# USB hub client

Install uhubctl and start `jazlabs-server-usb-hub --config default_lab --name usb_hub` first. Follow the physical mapping procedure in `docs/guides/usb-hub.md`.


```python
from JazLabs.hardware.USBHub.USBHub_Client import USBHubClient

HOST = "127.0.0.1"
COMMAND_PORT = 50951  # Local bridge: 51951.
```

```python
hub = USBHubClient(host=HOST, command_port=COMMAND_PORT)
status = hub.GetStatus()
print(status["trees"])
print(status["hubs"])
```
